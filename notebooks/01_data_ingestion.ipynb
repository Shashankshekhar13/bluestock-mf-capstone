# 01. Data Ingestion

This notebook documents the data ingestion phase of the Bluestock Mutual Fund Capstone Project. It verifies that all raw CSV files (Fund Master, NAV history, Transactions, Performance metrics, benchmark indices) are successfully loaded into the `data/raw/` directory.

```python
import os
import pandas as pd

raw_dir = '../data/raw'
if not os.path.exists(raw_dir):
    raw_dir = 'data/raw'

print("Listing raw files in ingestion directory:")
raw_files = [f for f in os.listdir(raw_dir) if f.endswith('.csv')]
for f in sorted(raw_files):
    path = os.path.join(raw_dir, f)
    size_kb = os.path.getsize(path) / 1024
    print(f"- {f} ({size_kb:.2f} KB)")
```

## Preview of raw master CSVs
Let's load a few samples of the raw datasets to verify columns.

```python
fund_master_path = os.path.join(raw_dir, '01_fund_master.csv')
if os.path.exists(fund_master_path):
    df = pd.read_csv(fund_master_path)
    print("--- Fund Master Ingested Preview ---")
    display(df.head(3))
else:
    print("Fund master file not found!")
```
