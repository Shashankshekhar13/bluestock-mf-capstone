### Executive Summary & Guide
This notebook presents a comprehensive **Exploratory Data Analysis (EDA)** of the mutual fund market, investor demographics, and geographic distributions based on the Bluestock database.

#### Key Highlights:
1. **Interactive Time-Series Analyses** using **Plotly** for NAV trends and monthly SIP inflows.
2. **Advanced Demographic Insights** using **Seaborn** and **Matplotlib** to understand investor distribution by age, gender, state, and city tier.
3. **Portfolio Correlation & Asset Allocation** to analyze stock-level holdings and fund return co-movements.
4. **10 Key Documented Findings** aligned with specific chart references.

All visualizations are exported to `reports/plots/` for inclusion in the final internship report.


```python
import os
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
import plotly.express as px
import plotly.io as pio

# Set rendering format for Plotly in Jupyter
pio.renderers.default = "notebook"

# Apply premium styling configurations
sns.set_theme(style="whitegrid")
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Helvetica', 'Arial', 'DejaVu Sans']
plt.rcParams['text.color'] = '#2b2b2b'
plt.rcParams['axes.labelcolor'] = '#2b2b2b'
plt.rcParams['xtick.color'] = '#4f4f4f'
plt.rcParams['ytick.color'] = '#4f4f4f'
plt.rcParams['grid.color'] = '#e2e8f0'
plt.rcParams['grid.alpha'] = 0.8
plt.rcParams['axes.edgecolor'] = '#cbd5e1'
plt.rcParams['axes.linewidth'] = 0.8
plt.rcParams['figure.facecolor'] = '#f8fafc'
plt.rcParams['axes.facecolor'] = '#ffffff'
plt.rcParams['figure.figsize'] = (12, 6.5)

# Premium color palette: Indigo (Primary), Teal (Accent), Coral (Secondary), Gold (Highlight)
COLORS = ['#2b3a8c', '#0d8a72', '#e66f50', '#e6ad12', '#7c3aed', '#2563eb', '#64748b']
sns.set_palette(sns.color_palette(COLORS))

# Ensure output directories exist
os.makedirs('../reports/plots', exist_ok=True)
os.makedirs('reports/plots', exist_ok=True)

# Connect to database
db_path = '../bluestock_mf.db'
if not os.path.exists(db_path):
    db_path = 'bluestock_mf.db'
    
print(f"Connecting to database: {os.path.abspath(db_path)}")
conn = sqlite3.connect(db_path)
```

## 1. NAV Trend Analysis (2022-2026)
Plotting the daily Net Asset Value (NAV) for all 40 mutual fund schemes. We highlight the **2023 Bull Run** and **2024 Market Correction** phases to analyze scheme resilience and performance.


```python
# Query NAV History joined with fund master details
df_nav = pd.read_sql_query("""
    SELECT n.date, n.nav, f.scheme_name, f.category, f.amfi_code
    FROM fact_nav n
    JOIN dim_fund f ON n.amfi_code = f.amfi_code
    ORDER BY n.date
""", conn)

df_nav['date'] = pd.to_datetime(df_nav['date'])

# Define key market event windows based on NIFTY50 index trend
# Bull run in 2023: June to Dec
# Market correction in 2024: March to October
bull_run_start, bull_run_end = '2023-06-01', '2023-12-31'
correction_start, correction_end = '2024-03-13', '2024-10-31'

# We'll create a Plotly interactive chart
fig = go.Figure()


highlight_schemes = {
    100016: ('HDFC Top 100 Fund (Large Cap)', '#2b3a8c'),
    118634: ('Nippon India Small Cap Fund', '#e66f50'),
    100025: ('HDFC Short Term Debt Fund', '#0d8a72')
}

unique_funds = df_nav['amfi_code'].unique()

for code in unique_funds:
    fund_data = df_nav[df_nav['amfi_code'] == code].sort_values('date')
    scheme_name = fund_data['scheme_name'].iloc[0]
    
    if code in highlight_schemes:
        name, color = highlight_schemes[code]
        fig.add_trace(go.Scatter(
            x=fund_data['date'], y=fund_data['nav'],
            mode='lines', name=name,
            line=dict(color=color, width=2.5)
        ))
    else:
        fig.add_trace(go.Scatter(
            x=fund_data['date'], y=fund_data['nav'],
            mode='lines', name=scheme_name,
            line=dict(color='rgba(148, 163, 184, 0.25)', width=0.8),
            showlegend=False
        ))

# Highlight the event windows
fig.add_vrect(
    x0=bull_run_start, x1=bull_run_end,
    fillcolor="rgba(16, 185, 129, 0.12)", opacity=0.9,
    layer="below", line_width=0,
    annotation_text="2023 Bull Run<br>(Jun - Dec)",
    annotation_position="top left",
    annotation_font=dict(size=11, color="#047857", family="Arial")
)

fig.add_vrect(
    x0=correction_start, x1=correction_end,
    fillcolor="rgba(239, 68, 68, 0.12)", opacity=0.9,
    layer="below", line_width=0,
    annotation_text="2024 Correction<br>(Mar - Oct)",
    annotation_position="top left",
    annotation_font=dict(size=11, color="#b91c1c", family="Arial")
)

fig.update_layout(
    title=dict(
        text="Daily NAV Trend Analysis (40 Schemes, 2022-2026)",
        font=dict(family="Arial", size=18, color="#1e293b")
    ),
    xaxis_title="Date",
    yaxis_title="Net Asset Value (NAV)",
    hovermode="x",
    template="plotly_white",
    legend=dict(
        orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1,
        font=dict(size=10)
    ),
    margin=dict(l=40, r=40, t=100, b=40)
)

fig.show()

# Generate Matplotlib equivalent static plot for exporting
plt.figure(figsize=(12, 6.5))
for code in unique_funds:
    fund_data = df_nav[df_nav['amfi_code'] == code].sort_values('date')
    scheme_name = fund_data['scheme_name'].iloc[0]
    
    if code in highlight_schemes:
        name, color = highlight_schemes[code]
        plt.plot(fund_data['date'], fund_data['nav'], label=name, color=color, linewidth=2.0, zorder=5)
    else:
        plt.plot(fund_data['date'], fund_data['nav'], color='#cbd5e1', linewidth=0.5, alpha=0.6, zorder=1)

# Shading for bull run and correction
plt.axvspan(pd.to_datetime(bull_run_start), pd.to_datetime(bull_run_end), color='#d1fae5', alpha=0.5, label='2023 Bull Run')
plt.axvspan(pd.to_datetime(correction_start), pd.to_datetime(correction_end), color='#fee2e2', alpha=0.5, label='2024 Market Correction')

plt.title('Daily NAV Trend Analysis (2022-2026)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Date', fontsize=11, labelpad=8)
plt.ylabel('Net Asset Value (NAV)', fontsize=11, labelpad=8)
plt.legend(loc='upper left', frameon=True, facecolor='#ffffff', edgecolor='#e2e8f0')
plt.tight_layout()
plt.savefig('reports/plots/01_nav_trends.png', dpi=300)
plt.savefig('../reports/plots/01_nav_trends.png', dpi=300)
plt.close()
```

## 2. AUM Growth & Fund House Comparison (2022-2025)
Grouped bar chart displaying the Assets Under Management (AUM) for each of the 10 fund houses. We highlight **SBI Mutual Fund's dominance**, which crossed the ₹12.5L Cr milestone in 2025.


```python
# Retrieve AUM history
df_aum = pd.read_sql_query("""
    SELECT date, fund_house, aum_lakh_crore, aum_crore, num_schemes
    FROM fact_aum
    ORDER BY date
""", conn)

df_aum['date'] = pd.to_datetime(df_aum['date'])
df_aum['year'] = df_aum['date'].dt.year

# Filter for the latest date in each calendar year to represent the year-end AUM
df_year_end = df_aum.loc[df_aum.groupby(['year', 'fund_house'])['date'].idxmax()]

# Sort fund houses by their average AUM to order legend consistently
order = df_year_end.groupby('fund_house')['aum_lakh_crore'].mean().sort_values(ascending=False).index

plt.figure(figsize=(13, 7.5))
# Define palette where SBI stands out in deep indigo, others in distinct colors
palette = {
    'SBI Mutual Fund': '#2b3a8c',
    'ICICI Prudential MF': '#0d8a72',
    'HDFC Mutual Fund': '#e66f50',
    'Nippon India MF': '#e6ad12',
    'Kotak Mahindra MF': '#7c3aed',
    'Axis Mutual Fund': '#2563eb',
    'UTI Mutual Fund': '#475569',
    'Aditya Birla Sun Life MF': '#94a3b8',
    'DSP Mutual Fund': '#cbd5e1',
    'Mirae Asset MF': '#e2e8f0'
}

sns.barplot(
    data=df_year_end, x='year', y='aum_lakh_crore', hue='fund_house',
    hue_order=order, palette=palette, edgecolor='none'
)

# Highlight SBI Mutual Fund at 12.5 Lakh Crore in 2025
sbi_2025_val = df_year_end[(df_year_end['year'] == 2025) & (df_year_end['fund_house'] == 'SBI Mutual Fund')]['aum_lakh_crore'].values[0]
plt.annotate(
    f"SBI Dominance: ₹{sbi_2025_val:.2f}L Cr\n(All-Time High)",
    xy=(3.0 - 0.35, sbi_2025_val), # Approximate x coordinate for SBI bar in 2025
    xytext=(1.8, 11.5),
    arrowprops=dict(facecolor='#2b3a8c', shrink=0.08, width=1.5, headwidth=7),
    fontsize=11.5, fontweight='bold', color='#1e293b',
    bbox=dict(boxstyle='round,pad=0.5', fc='#f8fafc', edgecolor='#2b3a8c', alpha=0.9)
)

plt.title('AUM Growth by Fund House & Year (2022-2025)', fontsize=15, fontweight='bold', pad=15)
plt.xlabel('Calendar Year', fontsize=11, labelpad=8)
plt.ylabel('Assets Under Management (₹ Lakh Crore)', fontsize=11, labelpad=8)
plt.legend(title='Asset Management Company (AMC)', bbox_to_anchor=(1.02, 1), loc='upper left', frameon=True)
plt.tight_layout()
plt.savefig('reports/plots/02_aum_growth.png', dpi=300)
plt.savefig('../reports/plots/02_aum_growth.png', dpi=300)
plt.show()
```

## 3. Monthly SIP Inflow Time-Series (2022-2025)
Monthly trend of Systematic Investment Plan (SIP) inflows across the entire industry. We annotate the **₹31,002 Cr record-high** achieved in December 2025 using Plotly.


```python
df_sip = pd.read_sql_query("""
    SELECT month, sip_inflow_crore, active_sip_accounts_crore, new_sip_accounts_lakh
    FROM fact_monthly_sip_inflows
    ORDER BY month
""", conn)

# Plotly line chart
fig = go.Figure()
fig.add_trace(go.Scatter(
    x=df_sip['month'], y=df_sip['sip_inflow_crore'],
    mode='lines+markers', name='SIP Inflow (Cr)',
    line=dict(color='#0d8a72', width=3),
    marker=dict(size=6, color='#2b3a8c')
))

# Annotate the peak in December 2025
peak_idx = df_sip['sip_inflow_crore'].idxmax()
peak_row = df_sip.iloc[peak_idx]

fig.add_annotation(
    x=peak_row['month'], y=peak_row['sip_inflow_crore'],
    text=f"All-Time High:<br>₹{peak_row['sip_inflow_crore']:,.0f} Cr (Dec 2025)",
    showarrow=True, arrowhead=2,
    ax=-100, ay=-50,
    font=dict(size=12, color="white"),
    bgcolor="#2b3a8c", bordercolor="#0d8a72", borderwidth=1.5,
    borderpad=5, opacity=0.95
)

fig.update_layout(
    title=dict(
        text="Monthly Industry SIP Inflows (Jan 2022 - Dec 2025)",
        font=dict(family="Arial", size=18, color="#1e293b")
    ),
    xaxis_title="Reporting Month (YYYY-MM)",
    yaxis_title="Monthly SIP Inflow (₹ Crore)",
    template="plotly_white"
)
fig.show()

# Matplotlib line plot for PNG export
plt.figure(figsize=(12, 6.5))
plt.plot(df_sip['month'], df_sip['sip_inflow_crore'], marker='o', color='#0d8a72', linewidth=2.5, markerfacecolor='#2b3a8c')
plt.xticks(df_sip['month'][::4], rotation=45) # Show label every 4 months

# Annotation
plt.annotate(
    f"All-Time High: ₹{peak_row['sip_inflow_crore']:,.0f} Cr\n(Dec 2025)",
    xy=(peak_idx, peak_row['sip_inflow_crore']),
    xytext=(peak_idx - 14, peak_row['sip_inflow_crore'] - 5000),
    arrowprops=dict(facecolor='#2b3a8c', shrink=0.08, width=1.5, headwidth=6),
    fontsize=11, fontweight='bold', color='#1e293b',
    bbox=dict(boxstyle='round,pad=0.4', fc='#f8fafc', edgecolor='#0d8a72', alpha=0.9)
)

plt.title('Monthly Mutual Fund Industry SIP Inflows (2022-2025)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Month', fontsize=11, labelpad=8)
plt.ylabel('SIP Inflow Amount (₹ Crore)', fontsize=11, labelpad=8)
plt.grid(True, which='both', linestyle='--', linewidth=0.5, color='#cbd5e1')
plt.tight_layout()
plt.savefig('reports/plots/03_sip_inflows.png', dpi=300)
plt.savefig('../reports/plots/03_sip_inflows.png', dpi=300)
plt.close()
```

## 4. Category Inflow Heatmap (Monthly Net Inflows)
Analyzing category-wise monthly net inflows to detect shifts in investor sentiment. Negative values indicate net redemptions (outflows).


```python
df_cat = pd.read_sql_query("""
    SELECT month, category, net_inflow_crore
    FROM fact_category_inflows
    ORDER BY month
""", conn)

# Pivot data to obtain category as rows and month as columns
df_pivot = df_cat.pivot(index='category', columns='month', values='net_inflow_crore')

plt.figure(figsize=(14, 6.5))
# Diverging colormap because net inflows contain negative numbers (outflows)
sns.heatmap(
    df_pivot, cmap='RdYlGn', center=0, annot=True, fmt=".0f",
    cbar_kws={'label': 'Net Inflow (₹ Crore)'}, linewidths=0.5,
    annot_kws={"size": 8}
)
plt.title('Monthly Net Inflows by Fund Category (₹ Crore)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Month (YYYY-MM)', fontsize=11, labelpad=8)
plt.ylabel('Fund Category', fontsize=11, labelpad=8)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('reports/plots/04_category_inflow_heatmap.png', dpi=300)
plt.savefig('../reports/plots/04_category_inflow_heatmap.png', dpi=300)
plt.show()
```

## 7. Folio Count Growth & Milestones (2022-2025)
Plotting the industrial folio count growth from **13.26 Cr** (Jan 2022) to **26.12 Cr** (Dec 2025). We mark key crossing milestones (15 Cr, 20 Cr, 25 Cr).


```python
df_folios = pd.read_sql_query("""
    SELECT month, total_folios_crore, equity_folios_crore, debt_folios_crore, hybrid_folios_crore
    FROM fact_industry_folio_count
    ORDER BY month
""", conn)

plt.figure(figsize=(12, 6.5))
plt.plot(df_folios['month'], df_folios['total_folios_crore'], color='#2b3a8c', marker='o', linewidth=3, label='Total Folios')
plt.plot(df_folios['month'], df_folios['equity_folios_crore'], color='#0d8a72', linestyle='--', linewidth=1.5, label='Equity Folios')

# Identify milestone months
milestones = [15.0, 20.0, 25.0]
milestone_points = []
for m in milestones:
    match = df_folios[df_folios['total_folios_crore'] >= m]
    if not match.empty:
        milestone_points.append(match.iloc[0])

# Annotate milestones
for pt in milestone_points:
    idx = df_folios[df_folios['month'] == pt['month']].index[0]
    plt.axvline(x=pt['month'], color='#e66f50', linestyle=':', alpha=0.8, linewidth=1.5)
    plt.plot(pt['month'], pt['total_folios_crore'], marker='X', markersize=10, color='#e66f50')
    plt.annotate(
        f"{pt['total_folios_crore']:.1f} Cr\n({pt['month']})",
        xy=(pt['month'], pt['total_folios_crore']),
        xytext=(idx - 2, pt['total_folios_crore'] + 1.2),
        fontsize=9.5, fontweight='bold', color='#e66f50',
        bbox=dict(boxstyle='round,pad=0.3', fc='#fff5f5', edgecolor='#e66f50', alpha=0.9)
    )

plt.title('Mutual Fund Folio Growth & Milestones (Jan 2022 - Dec 2025)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Month', fontsize=11, labelpad=8)
plt.ylabel('Accounts / Folios (in Crore)', fontsize=11, labelpad=8)
plt.xticks(df_folios['month'][::4], rotation=45)
plt.legend(loc='upper left', frameon=True)
plt.grid(True, which='both', linestyle='--', linewidth=0.5, color='#cbd5e1')
plt.ylim(10, 30)
plt.tight_layout()
plt.savefig('reports/plots/10_folio_growth.png', dpi=300)
plt.savefig('../reports/plots/10_folio_growth.png', dpi=300)
plt.show()
```

## 8. NAV Return Correlation Matrix (10 Selected Funds)
Compute and visualize the pairwise correlation matrix of daily returns for 10 representative funds (5 Equity, 5 Debt) to test diversification properties.


```python
# Load 5 Equity and 5 Debt funds dynamically
eq_funds = pd.read_sql_query("SELECT amfi_code, scheme_name FROM dim_fund WHERE category = 'Equity' LIMIT 5", conn)
debt_funds = pd.read_sql_query("SELECT amfi_code, scheme_name FROM dim_fund WHERE category = 'Debt' LIMIT 5", conn)
selected = pd.concat([eq_funds, debt_funds])
selected_codes = selected['amfi_code'].tolist()

# Query NAV history for selected schemes
df_corr_nav = pd.read_sql_query(f"""
    SELECT n.date, n.nav, f.scheme_name
    FROM fact_nav n
    JOIN dim_fund f ON n.amfi_code = f.amfi_code
    WHERE f.amfi_code IN ({','.join(map(str, selected_codes))})
    ORDER BY n.date
""", conn)

# Pivot and compute returns
df_corr_pivot = df_corr_nav.pivot(index='date', columns='scheme_name', values='nav')
df_returns = df_corr_pivot.pct_change().dropna()

# Shorten names for clean visualization
df_returns.columns = [col.split(' - ')[0][:30] for col in df_returns.columns]

# Compute correlation
corr_matrix = df_returns.corr()

plt.figure(figsize=(11, 9))
sns.heatmap(
    corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt=".2f",
    linewidths=0.5, cbar_kws={'label': 'Correlation Coefficient'}
)
plt.title('Daily Returns Correlation Heatmap (Equity vs Debt)', fontsize=14, fontweight='bold', pad=15)
plt.tight_layout()
plt.savefig('reports/plots/11_nav_correlation_matrix.png', dpi=300)
plt.savefig('../reports/plots/11_nav_correlation_matrix.png', dpi=300)
plt.show()
```

## 9. Sector Allocation Donut Chart (Equity Funds)
Aggregating stock-level holdings weights from the equity funds portfolio to examine sector exposures.


```python
df_holdings = pd.read_sql_query("""
    SELECT ph.sector, SUM(ph.weight_pct) as sum_weight
    FROM fact_portfolio_holdings ph
    JOIN dim_fund f ON ph.amfi_code = f.amfi_code
    WHERE f.category = 'Equity'
    GROUP BY ph.sector
    ORDER BY sum_weight DESC
""", conn)

# Normalize weights
df_holdings['weight_pct'] = (df_holdings['sum_weight'] / df_holdings['sum_weight'].sum()) * 100

# Select top 7 sectors, group others
top_n = 7
df_top = df_holdings.head(top_n).copy()
others_weight = df_holdings['weight_pct'].iloc[top_n:].sum()
df_top.loc[len(df_top)] = ['Others', df_holdings['sum_weight'].iloc[top_n:].sum(), others_weight]

plt.figure(figsize=(8, 8))
plt.pie(
    df_top['weight_pct'], labels=df_top['sector'], autopct='%1.1f%%', startangle=140,
    colors=COLORS, wedgeprops=dict(width=0.4, edgecolor='w', linewidth=2)
)
plt.title('Aggregate Sector Allocation Across Equity Portfolios', fontsize=14, fontweight='bold', pad=15)
plt.tight_layout()
plt.savefig('reports/plots/12_sector_allocation_donut.png', dpi=300)
plt.savefig('../reports/plots/12_sector_allocation_donut.png', dpi=300)
plt.show()
```

```python
# 13. Risk Category Distribution
df_risk = pd.read_sql_query("SELECT risk_category FROM dim_fund", conn)

plt.figure(figsize=(11, 5.5))
order_risk = ['Low', 'Low to Moderate', 'Moderate', 'Moderately High', 'High', 'Very High']
# We check if categories exist in our DB
existing_risk = [r for r in order_risk if r in df_risk['risk_category'].unique()]
sns.countplot(data=df_risk, x='risk_category', order=existing_risk, palette='coolwarm')
plt.title('Distribution of Mutual Fund Schemes by Risk Category', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Risk Category (AMFI Riskometer)', fontsize=11)
plt.ylabel('Scheme Count', fontsize=11)
plt.tight_layout()
plt.savefig('reports/plots/13_risk_distribution.png', dpi=300)
plt.savefig('../reports/plots/13_risk_distribution.png', dpi=300)
plt.show()

# 14. Top 10 Stock Holdings
df_stocks = pd.read_sql_query("""
    SELECT stock_name, sector, SUM(weight_pct) as total_weight
    FROM fact_portfolio_holdings
    GROUP BY stock_name, sector
    ORDER BY total_weight DESC
    LIMIT 10
""", conn)

plt.figure(figsize=(11, 6.5))
sns.barplot(data=df_stocks, x='total_weight', y='stock_name', hue='sector', palette='Set1', dodge=False)
plt.title('Top 10 Stocks by Aggregated Portfolio Weight (%)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Sum of Portfolio Weight (%)', fontsize=11)
plt.ylabel('Stock Name', fontsize=11)
plt.legend(title='Sector', loc='lower right')
plt.tight_layout()
plt.savefig('reports/plots/14_top_stocks.png', dpi=300)
plt.savefig('../reports/plots/14_top_stocks.png', dpi=300)
plt.show()

# 15. Expense Ratio vs Returns
df_exp = pd.read_sql_query("""
    SELECT scheme_name, category, expense_ratio_pct, return_3yr_pct
    FROM fact_performance
    WHERE return_3yr_pct IS NOT NULL AND expense_ratio_pct IS NOT NULL
""", conn)

plt.figure(figsize=(11, 6.5))
sns.scatterplot(
    data=df_exp, x='expense_ratio_pct', y='return_3yr_pct', hue='category',
    palette=['#2b3a8c', '#e66f50'], style='category', s=80, alpha=0.9
)
# Annotate outliers
top_ret = df_exp.sort_values('return_3yr_pct', ascending=False).iloc[0]
plt.annotate(
    top_ret['scheme_name'][:30],
    xy=(top_ret['expense_ratio_pct'], top_ret['return_3yr_pct']),
    xytext=(top_ret['expense_ratio_pct'] + 0.1, top_ret['return_3yr_pct'] - 2),
    arrowprops=dict(arrowstyle="->", color="black", connectionstyle="arc3,rad=.2"),
    fontsize=9
)
plt.title('Expense Ratio (%) vs 3-Year Annualised Returns (%)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Expense Ratio (%)', fontsize=11)
plt.ylabel('3-Year Annualised Return (%)', fontsize=11)
plt.tight_layout()
plt.savefig('reports/plots/15_expense_ratio_vs_returns.png', dpi=300)
plt.savefig('../reports/plots/15_expense_ratio_vs_returns.png', dpi=300)
plt.show()

# 16. Benchmark Index Performance (Plotly & static)
df_indices = pd.read_sql_query("""
    SELECT date, index_name, close_value
    FROM fact_benchmark_indices
    WHERE index_name IN ('NIFTY50', 'NIFTY_MIDCAP150', 'BSE_SMALLCAP', 'CRISIL_GILT')
    ORDER BY date
""", conn)

df_indices['date'] = pd.to_datetime(df_indices['date'])

# Normalize close values to base 100 on the first date for comparison
df_norm = []
for idx in df_indices['index_name'].unique():
    subset = df_indices[df_indices['index_name'] == idx].sort_values('date').copy()
    base_val = subset['close_value'].iloc[0]
    subset['norm_value'] = (subset['close_value'] / base_val) * 100
    df_norm.append(subset)
df_norm = pd.concat(df_norm)

fig_idx = px.line(
    df_norm, x='date', y='norm_value', color='index_name',
    labels={'norm_value': 'Normalized Index Value (Base 100)', 'date': 'Date'},
    title='Benchmark Indices Relative Performance (Base 100, 2022-2026)',
    color_discrete_sequence=['#2b3a8c', '#e66f50', '#0d8a72', '#e6ad12']
)
fig_idx.update_layout(template="plotly_white")
fig_idx.show()

# Matplotlib equivalent for static export
plt.figure(figsize=(12, 6.5))
for idx in df_norm['index_name'].unique():
    subset = df_norm[df_norm['index_name'] == idx].sort_values('date')
    plt.plot(subset['date'], subset['norm_value'], label=idx, linewidth=2.0)

plt.title('Benchmark Indices Relative Performance (2022-2026, Base 100)', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Date', fontsize=11)
plt.ylabel('Normalized Value (Base 100)', fontsize=11)
plt.legend(frameon=True)
plt.tight_layout()
plt.savefig('reports/plots/16_benchmark_comparison.png', dpi=300)
plt.savefig('../reports/plots/16_benchmark_comparison.png', dpi=300)
plt.close()
```
