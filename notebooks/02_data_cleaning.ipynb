## Data Quality Check: AMFI Codes
Verify that all AMFI codes inside the NAV History file successfully match with the codes defined in the Fund Master dimension table.

```python
fm = pd.read_csv(os.path.join(raw_dir, '01_fund_master.csv'))
nav = pd.read_csv(os.path.join(raw_dir, '02_nav_history.csv'))

fm_codes = set(fm['amfi_code'].unique())
nav_codes = set(nav['amfi_code'].unique())

matched = fm_codes.intersection(nav_codes)
print(f"Unique codes in Fund Master: {len(fm_codes)}")
print(f"Unique codes in NAV History: {len(nav_codes)}")
print(f"Matched unique codes: {len(matched)} ({len(matched)/len(fm_codes)*100:.2f}%)")
```
